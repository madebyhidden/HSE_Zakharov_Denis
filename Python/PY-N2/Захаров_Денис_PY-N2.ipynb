```python
def date_range (start_date:str, end_date:str) -> list:
    answer = []
    format_ = "%Y-%m-%d"
    start_date, end_date = dt.datetime.strptime(start_date, format_), dt.datetime.strptime(end_date, format_)
    if end_date < start_date:
        return answer
    else:
        answer.append(start_date.strftime(format_))
        timedelta_ = end_date - start_date
        for i in range(0, int(str(timedelta_).split(" ")[0])):
            start_date += dt.timedelta(days=1)
            answer.append(start_date.strftime(format_))
        return answer   
    
date_range('2022-01-01', '2022-01-03')
```
