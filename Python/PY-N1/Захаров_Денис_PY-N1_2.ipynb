# Задание 1

```python
import json
purchases = {}
i = 1

with open("purchase_log.txt", "r", encoding="cp1251") as file:
    file.__next__()
    a = file.readlines()

for element in a:
    element_js = json.loads(element)
    value = element_js["category"]
    key = element_js["user_id"]
    purchases.update({key : value})
 
for key, value in purchases.items():
    if i == 3: 
        break
    print(key, value)
    i += 1
    
```

```python
pd.DataFrame.from_dict(purchases, orient='index').head()
```

# Задание 2

```python
with open("funnel.csv", "w", encoding="cp1251") as end_file, open("visit_log (2).csv", "r", encoding="cp1251") as file_2:
        end_file.write(f"{file_2.readline().replace("\n", "")},category\n")
        file_2.__next__() 
        for line in file_2:
            if line.split(",")[0] in purchases.keys():
                end_file.write(f"{line.replace("\n", "")},{purchases[line.split(",")[0]]}\n")
                
        
```
