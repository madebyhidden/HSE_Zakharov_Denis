```python
import re
```

# Задание 1

```python
allowed_letters = 'ABEKMOPCTYX'
```

```python
pattern = re.compile(r'^([' + allowed_letters + r']\d{3}[' + allowed_letters + r']{2})(\d{2,3})$')
```

```python
print(pattern)
```

```python
if pattern.match('A222BC96'):
    match = pattern.match('A222BC96')
  
    number_part = match.group(1)
    region_part = match.group(2)
    print(f'Номер {number_part} валиден. Регион: {region_part}.')
else:
    print('Номер не валиден.')
```

```python
if pattern.match('АБ22ВВ193'):
    match = pattern.match('АБ22ВВ193')

    number_part = match.group(1)
    region_part = match.group(2)
    print(f'Номер {number_part} валиден. Регион: {region_part}.')
else:
    print('Номер не валиден.')
```

```python
if pattern.match('АВ22В193'):   # Используется кириллица
    match = pattern.match('АВ22В193')

    number_part = match.group(1)
    region_part = match.group(2)
    print(f'Номер {number_part} валиден. Регион: {region_part}.')
else:
    print('Номер не валиден.')
```

# Задание 2

```python
text = 'Напишите функцию функцию, которая будет будет будет будет удалять все все все все последовательные повторы слов из из из из заданной строки строки при помощи регулярных выражений'
```

```python
result = re.sub(r'\b(\w+)(?:\s+\1\b)+', r'\1', text, flags=re.IGNORECASE)
```

```python
print(result)
```
